```python
# ============================================================
# Notebook 01: Dataset Understanding
# Project: Financial Scam Narrative Detection and Fraud Analysis
# ============================================================

# Data manipulation
import pandas as pd
import numpy as np

# Data visualization
import matplotlib.pyplot as plt
import seaborn as sns

# File and directory management
from pathlib import Path

# Display configuration
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", 100)

# Visualization configuration
sns.set_theme(style="whitegrid", palette="deep")
plt.rcParams["figure.figsize"] = (10, 6)

print("Libraries imported successfully!")
```

```python
# 1. Project directory configuration

# Current notebook working directory
NOTEBOOK_DIR = Path.cwd()

# Project root directory
PROJECT_ROOT = NOTEBOOK_DIR.parent

# Dataset directory
DATA_DIR = PROJECT_ROOT / "data" / "splits"

# Dataset file paths
TRAIN_PATH = DATA_DIR / "train.csv"
VALIDATION_PATH = DATA_DIR / "validation.csv"
TEST_PATH = DATA_DIR / "test.csv"

print("Project root:", PROJECT_ROOT)
print("Dataset directory:", DATA_DIR)

print("\nDataset availability:")
print("Train:", TRAIN_PATH.exists())
print("Validation:", VALIDATION_PATH.exists())
print("Test:", TEST_PATH.exists())
```

```python
# 2. Load the training dataset

train_df = pd.read_csv(TRAIN_PATH, low_memory=False)

print("Training dataset loaded successfully!")
print("Shape:", train_df.shape)
```

```python
# 3. Inspect dataset structure

print("COLUMN NAMES:\n")

for index, column in enumerate(train_df.columns, start=1):
    print(f"{index:02d}. {column}")

print("\nDATA TYPES:\n")
print(train_df.dtypes)

print("\nDATASET INFORMATION:\n")
train_df.info()
```

```python
# 4. Preview dataset records

train_df.head()
```

```python
# 5. Compare representative scam and legitimate messages

scam_example = train_df[train_df["label"] == 1].iloc[0]
legit_example = train_df[train_df["label"] == 0].iloc[0]

print("SCAM MESSAGE")
print("-" * 60)
print(scam_example["normalized_text"][:1500])

print("\n\nLEGITIMATE MESSAGE")
print("-" * 60)
print(legit_example["normalized_text"][:1500])
```

```python
# 6. Analyze class distribution

class_counts = train_df["label"].value_counts().sort_index()

class_percentages = train_df["label"].value_counts(normalize=True).sort_index() * 100

class_distribution = pd.DataFrame({
    "Count": class_counts,
    "Percentage": class_percentages
})

class_distribution.index = ["Legitimate (0)", "Scam (1)"]

class_distribution
```

```python
# 7. Visualize class distribution

plt.figure(figsize=(8, 5))

sns.barplot(
    x=["Legitimate", "Scam"],
    y=class_counts.values
)

plt.title("Class Distribution in Training Dataset")
plt.xlabel("Message Class")
plt.ylabel("Number of Messages")

plt.show()
```

```python
# 8. Analyze channel distribution

channel_distribution = pd.crosstab(
    train_df["label"],
    train_df["channel"]
)

channel_distribution.index = ["Legitimate (0)", "Scam (1)"]

channel_distribution
```

```python
# 9. Calculate channel percentages by class

channel_percentages = pd.crosstab(
    train_df["label"],
    train_df["channel"],
    normalize="index"
) * 100

channel_percentages.index = ["Legitimate (0)", "Scam (1)"]

channel_percentages.round(2)
```

```python
# 10. Calculate class distribution within each channel

class_by_channel = pd.crosstab(
    train_df["channel"],
    train_df["label"],
    normalize="index"
) * 100

class_by_channel.columns = ["Legitimate (0)", "Scam (1)"]

class_by_channel.round(2)
```
