```python
# 23. Analyze most frequent placeholder tokens

import re
from collections import Counter

placeholder_pattern = r"\[[A-Z_]+\]"

all_placeholders = []

for text in train_df["normalized_text"]:
    all_placeholders.extend(
        re.findall(placeholder_pattern, text)
    )

placeholder_frequency = pd.Series(
    Counter(all_placeholders)
).sort_values(ascending=False)

placeholder_frequency.head(20)
```

```python
# 24. Compare placeholder frequencies by class

placeholder_class_counts = {}

for label, group in train_df.groupby("label"):
    counter = Counter()

    for text in group["normalized_text"]:
        counter.update(re.findall(placeholder_pattern, text))

    placeholder_class_counts[label] = counter

placeholder_class_df = pd.DataFrame(
    placeholder_class_counts
).fillna(0).astype(int)

placeholder_class_df.columns = ["Legitimate (0)", "Scam (1)"]

placeholder_class_df["Total"] = (
    placeholder_class_df["Legitimate (0)"] +
    placeholder_class_df["Scam (1)"]
)

placeholder_class_df = placeholder_class_df.sort_values(
    "Total", ascending=False
)

placeholder_class_df.head(20)
```

```python
# 25. Calculate placeholder presence by class

document_placeholder_counts = {}

for label, group in train_df.groupby("label"):
    counter = Counter()

    for text in group["normalized_text"]:
        unique_tokens = set(re.findall(placeholder_pattern, text))
        counter.update(unique_tokens)

    document_placeholder_counts[label] = counter

document_placeholder_df = pd.DataFrame(
    document_placeholder_counts
).fillna(0).astype(int)

document_placeholder_df.columns = ["Legitimate (0)", "Scam (1)"]

document_placeholder_df["Legitimate (%)"] = (
    document_placeholder_df["Legitimate (0)"] /
    (train_df["label"] == 0).sum() * 100
)

document_placeholder_df["Scam (%)"] = (
    document_placeholder_df["Scam (1)"] /
    (train_df["label"] == 1).sum() * 100
)

document_placeholder_df = document_placeholder_df.sort_values(
    "Total" if "Total" in document_placeholder_df.columns else "Legitimate (0)",
    ascending=False
)

document_placeholder_df.round(2)
```

```python
# 26. Analyze frequent words by class

from collections import Counter
import re

word_frequencies = {}

for label, group in train_df.groupby("label"):
    counter = Counter()

    for text in group["normalized_text"]:
        text = re.sub(r"\[[^\]]*\]", " ", text)
        words = re.findall(r"\b[a-zA-Z]{2,}\b", text.lower())
        counter.update(words)

    word_frequencies[label] = counter

print("Most frequent legitimate words:")
print(word_frequencies[0].most_common(30))

print("\nMost frequent scam words:")
print(word_frequencies[1].most_common(30))
```

```python
# 27. Load English stopwords

import nltk
from nltk.corpus import stopwords

nltk.download("stopwords")

stop_words = set(stopwords.words("english"))

print("Total English stopwords:", len(stop_words))
print("\nSample stopwords:")
print(sorted(stop_words)[:30])
```

```python
# 28. Compare frequent words against stopwords

for label, class_name in [(0, "Legitimate"), (1, "Scam")]:
    top_words = word_frequencies[label].most_common(30)

    stopword_matches = [
        (word, count) for word, count in top_words
        if word in stop_words
    ]

    informative_words = [
        (word, count) for word, count in top_words
        if word not in stop_words
    ]

    print(f"\n{class_name}")
    print("Top stopwords:", stopword_matches)
    print("Top non-stopwords:", informative_words)
```

```python
# 29. Explore frequent non-stopwords by class

for label, class_name in [(0, "Legitimate"), (1, "Scam")]:
    filtered_words = {
        word: count
        for word, count in word_frequencies[label].items()
        if word not in stop_words
    }

    print(f"\n{class_name} — Top 30 non-stopwords")
    print(Counter(filtered_words).most_common(30))
```

```python
# 30. Compare normalized word frequencies by class

total_words = {
    label: sum(counter.values())
    for label, counter in word_frequencies.items()
}

comparison_words = [
    "account", "bank", "money", "transaction",
    "transfer", "email", "file", "code",
    "please", "information", "security"
]

rows = []

for word in comparison_words:
    legit_rate = (
        word_frequencies[0][word] / total_words[0]
    ) * 10000

    scam_rate = (
        word_frequencies[1][word] / total_words[1]
    ) * 10000

    rows.append({
        "word": word,
        "legitimate_per_10k": round(legit_rate, 2),
        "scam_per_10k": round(scam_rate, 2)
    })

frequency_comparison = pd.DataFrame(rows)

display(frequency_comparison)
```
